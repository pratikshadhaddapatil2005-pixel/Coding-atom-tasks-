```python
# STEP 7: ERROR HANDLING AND RETRIES

import requests
import time

def extract_weather_data(latitude, longitude, max_retries=3):
    url = (
        "https://api.open-meteo.com/v1/forecast"
        f"?latitude={latitude}"
        f"&longitude={longitude}"
        "&current=temperature_2m,relative_humidity_2m,"
        "wind_speed_10m,weather_code"
    )

    for attempt in range(1, max_retries + 1):
        try:
            print(f"🔄 Attempt {attempt}...")

            response = requests.get(url, timeout=10)
            response.raise_for_status()

            print("✅ Data extracted successfully!")
            return response.json()

        except requests.exceptions.RequestException as error:
            print(f"❌ Attempt {attempt} failed: {error}")

            if attempt < max_retries:
                print("⏳ Retrying in 2 seconds...")
                time.sleep(2)
            else:
                print("❌ All retry attempts failed.")
                return None


# Test the function
new_data = extract_weather_data(12.9716, 77.5946)

if new_data is not None:
    print("🎉 Extraction completed successfully!")
else:
    print("⚠️ Extraction failed.")
```

```python

# ============================================================
# WEATHER DATA ETL PIPELINE
# Extract → Transform → Quality Check → Load
# With Retry + Logging
# ============================================================

import requests
import pandas as pd
import sqlite3
import logging
import time

# ---------------- CONFIGURATION ----------------

CITY = "Bengaluru"
LATITUDE = 12.9716
LONGITUDE = 77.5946

DATABASE = "weather_data.db"
TABLE_NAME = "weather_data"
LOG_FILE = "etl_pipeline.log"

# ---------------- LOGGING ----------------

logging.basicConfig(
    filename=LOG_FILE,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

# ---------------- EXTRACT ----------------

def extract_weather_data(max_retries=3):

    url = (
        "https://api.open-meteo.com/v1/forecast"
        f"?latitude={LATITUDE}"
        f"&longitude={LONGITUDE}"
        "&current=temperature_2m,"
        "relative_humidity_2m,"
        "wind_speed_10m,"
        "weather_code"
    )

    for attempt in range(1, max_retries + 1):

        try:
            print(f"🔄 Extraction attempt {attempt}...")
            logging.info(f"Extraction attempt {attempt}")

            response = requests.get(url, timeout=10)
            response.raise_for_status()

            data = response.json()

            print("✅ Data extracted successfully")
            logging.info("Weather data extracted successfully")

            return data

        except requests.exceptions.RequestException as error:

            print(f"❌ Attempt {attempt} failed")
            logging.error(f"Attempt {attempt} failed: {error}")

            if attempt < max_retries:
                print("⏳ Retrying in 2 seconds...")
                time.sleep(2)

            else:
                print("❌ All extraction attempts failed")
                logging.error("All extraction attempts failed")

                return None


# ---------------- TRANSFORM ----------------

def transform_data(data):

    current = data["current"]

    weather_data = {
        "city": CITY,
        "temperature": current["temperature_2m"],
        "humidity": current["relative_humidity_2m"],
        "wind_speed": current["wind_speed_10m"],
        "weather_code": current["weather_code"],
        "timestamp": current["time"]
    }

    df = pd.DataFrame([weather_data])

    print("🔄 Data transformed successfully")
    logging.info("Data transformation completed")

    return df


# ---------------- DATA QUALITY ----------------

def validate_data(df):

    print("\n🔍 Running data quality checks...")

    # Missing values
    if df.isnull().sum().sum() > 0:
        raise ValueError("Missing values found")

    print("✅ No missing values")

    # Temperature
    if not df["temperature"].between(-50, 60).all():
        raise ValueError("Invalid temperature")

    print("✅ Temperature is valid")

    # Humidity
    if not df["humidity"].between(0, 100).all():
        raise ValueError("Invalid humidity")

    print("✅ Humidity is valid")

    # Wind speed
    if (df["wind_speed"] < 0).any():
        raise ValueError("Invalid wind speed")

    print("✅ Wind speed is valid")

    # Duplicate check
    if df.duplicated().any():
        raise ValueError("Duplicate record found")

    print("✅ No duplicate records")

    logging.info("All data quality checks passed")

    return True


# ---------------- LOAD ----------------

def load_data(df):

    conn = sqlite3.connect(DATABASE)

    df.to_sql(
        TABLE_NAME,
        conn,
        if_exists="append",
        index=False
    )

    conn.close()

    print("🗄️ Data loaded into SQLite database")
    logging.info("Data loaded into SQLite database")


# ---------------- MAIN ETL PIPELINE ----------------

def run_etl():

    print("\n===================================")
    print("🚀 WEATHER ETL PIPELINE STARTED")
    print("===================================\n")

    logging.info("ETL pipeline started")

    try:

        # Extract
        data = extract_weather_data()

        if data is None:
            raise Exception("Extraction failed")

        # Transform
        df = transform_data(data)

        print("\n📊 Transformed Data:")
        print(df)

        # Data Quality
        validate_data(df)

        # Load
        load_data(df)

        print("\n===================================")
        print("🎉 ETL PIPELINE COMPLETED")
        print("===================================")

        logging.info("ETL pipeline completed successfully")

    except Exception as error:

        print("\n❌ ETL PIPELINE FAILED")
        print("Error:", error)

        logging.error(f"ETL pipeline failed: {error}")


# ---------------- RUN PIPELINE ----------------

run_etl()
```
